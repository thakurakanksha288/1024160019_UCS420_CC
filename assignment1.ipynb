**Assignment 5.3:** WAP to add all prime numbers from 1 to n, where n is given by the user.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter n: "))
prime_sum = sum(i for i in range(1, n + 1) if is_prime(i))
print("Sum of primes 1 to", n, "-->", prime_sum)
```

**Assignment 6.2:** WAP using a function that adds all prime numbers from 1 to n, where n is given by the user.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

def sum_primes(n):
    return sum(i for i in range(1, n + 1) if is_prime(i))

n = int(input("Enter n: "))
print("Sum of primes 1 to", n, "-->", sum_primes(n))
```

## 10. Exception Handling
**Learning:** Catching runtime errors with try/except.

```python
values = [10, 5, 0, 2]
for v in values:
    try:
        print("100 /", v, "-->", 100 / v)
    except ZeroDivisionError:
        print("100 /", v, "--> error: division by zero")
```

## 11. Data Structure 1 — List

```python
L = [1024, 1601, 9, 10, 24]
print("Original list -->", L)

L.append(2026)
print("After append  -->", L)

del L[1]
print("After delete  -->", L)

print("Sum     -->", sum(L))
print("Average -->", sum(L) / len(L))

L.sort()
print("Ascending  -->", L)
L.sort(reverse=True)
print("Descending -->", L)

print("9 in L?", 9 in L)
```

```python
cgpa = {1: 8.9, 2: 5.6, 4: 6.7, 7: 9.1, 8: 5.3}
print("Dictionary -->", cgpa)
print("CGPA of 4  -->", cgpa[4])

cgpa[4] = 9.2          # update
cgpa[3] = 8.6          # add
print("After update/add -->", cgpa)

del cgpa[1]            # delete
print("After delete     -->", cgpa)

print("Is key 7 present?", 7 in cgpa)

for k, v in cgpa.items():
    print("Roll", k, "-> CGPA", v)
```

## 13. Data Structure 3 — Tuple
**Learning:** Declaring, iterating, and searching a tuple; why tuples are immutable.

---

```python
T = ("Cognitive", "Computing", 3.14, 3)
print("T               -->", T)
print("Num of elements -->", len(T))

for item in T:
    print(item)

print("T[1:3] -->", T[1:3])
print("Max/Min of (3,6,9,12,5,3,2) -->", max((3, 6, 9, 12, 5, 3, 2)), min((3, 6, 9, 12, 5, 3, 2)))

# Tuples are immutable, so there's no in-place add/insert/delete.
# The common workaround ("jugaad") is to convert to a list, edit it,
# then convert back to a tuple:
T_list = list(T)
T_list.append("Extra")
T = tuple(T_list)
print("After append (via list) -->", T)

try:
    T[0] = "Changed"
except TypeError as e:
    print("Direct assignment fails:", e)
```

## 14. Data Structure 4 — Set
**Learning:** Declaring a set and set operations (union, intersection, difference).

---

```python
a = {1, 2, 4, 7, 8}
b = {2, 3, 4, 9}
print("Set a -->", a)
print("Set b -->", b)

print("Union        -->", a | b)
print("Intersection -->", a & b)
print("a - b        -->", a - b)
print("b - a        -->", b - a)

a.add(99)
print("After add(99)    -->", a)
a.discard(99)
print("After discard(99)-->", a)
```
