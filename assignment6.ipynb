```python
import numpy as np
#question-1

# Original sensor readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
temp_corrected = temperature + 2
print("a) Corrected Readings (+2°C):", temp_corrected)

# b. Convert all temperatures from Celsius to Fahrenheit
temp_fahrenheit = (9 / 5) * temp_corrected + 32
print("b) Temperatures in Fahrenheit:", temp_fahrenheit)

# c. Identify all readings greater than 32°C using Boolean indexing
readings_above_32 = temp_corrected[temp_corrected > 32]
print("c) Readings > 32°C:", readings_above_32)

# d. Count how many readings exceed 32°C
count_above_32 = np.sum(temp_corrected > 32)
print("d) Number of readings > 32°C:", count_above_32)
```

```python
#question - 2

# Daily steps matrix (4 users x 3 days)
steps = np.array(
    [
        [5000, 6200, 7100],
        [8000, 7500, 9000],
        [4500, 5100, 4800],
        [9000, 8500, 9500],
    ]
)

# a. Total steps recorded
total_steps = np.sum(steps)
print("a) Total steps recorded:", total_steps)

# b. Mean number of steps
mean_steps = np.mean(steps)
print("b) Mean number of steps:", mean_steps)

# c. Maximum and minimum values
max_val = np.max(steps)
min_val = np.min(steps)
print(f"c) Maximum steps: {max_val} | Minimum steps: {min_val}")

# d. Total steps for each day (sum across rows / along axis 0)
daily_totals = np.sum(steps, axis=0)
print("d) Total steps for each day (Day 1, 2, 3):", daily_totals)

# e. Total steps for each user (sum across columns / along axis 1)
user_totals = np.sum(steps, axis=1)
print("e) Total steps for each user (User 1, 2, 3, 4):", user_totals)

# f. Position of maximum steps using argmax()
flat_index = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_index, steps.shape)

print(f"f) Flattened argmax index: {flat_index}")
print(
    f"   2D Matrix Position (User, Day): Row index {user_idx}, Column index {day_idx}"
)
print(f"   (User {user_idx + 1}, Day {day_idx + 1})")
```

```python
#question-3
# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a) Original array:", original)

# b. Create slice containing elements from index 1 to 4 (indices 1, 2, 3, 4)
subset = original[1:5]
print("b) Subset slice:", subset)

# c. Modify first element of subset to 999
subset[0] = 999
print("c) After modifying subset[0] = 999:")
print("   original:", original)
print("   subset:", subset)

# d. Create slice with .copy() and modify
copied = original[1:5].copy()
copied[0] = 500
print("d) After modifying copied[0] = 500:")
print("   original:", original)
print("   copied:", copied)

# e. Create 1 to 12 array and reshape to 3 x 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne) 3 x 4 Matrix:\n", matrix)

# f. Slicing operations on 3 x 4 matrix
print("\nf) Indexing and Slicing:")
print("   First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Elements from rows 1-2 & columns 2-3:\n", matrix[1:3, 2:4])

# g. Flattening matrix using flatten() and ravel()
flat_arr = matrix.flatten()
ravel_arr = matrix.ravel()
print("\ng) Flattened array (flatten()):", flat_arr)
print("   Flattened array (ravel()):", ravel_arr)

# h. Modify element in array returned by ravel()
ravel_arr[0] = 111
print("\nh) After modifying ravel_arr[0] = 111:")
print("   Original 3x4 Matrix (affected): \n", matrix)

# Reset matrix
matrix = np.arange(1, 13).reshape(3, 4)

# i. Modify element in array returned by flatten()
flat_arr = matrix.flatten()
flat_arr[0] = 222
print("\ni) After modifying flat_arr[0] = 222:")
print("   Original 3x4 Matrix (unaffected): \n", matrix)

# j. Display attributes of 3 x 4 matrix
print("\nj) Array Attributes:")
print("   Shape:", matrix.shape)
print("   Dimensions (ndim):", matrix.ndim)
print("   Total elements (size):", matrix.size)
print("   Data type (dtype):", matrix.dtype)
```

```python
#question-4
# Target vector (Assistance score)
y = np.array([40, 65, 30, 85])

# Feature matrix (Sleep hours, Activity level, Stress level)
X = np.array([[6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8]])

# a. Shape and dimensions of X
print("a) Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

# b. Calculate X.T (Transpose)
X_T = X.T
print("\nb) Transpose of X (X.T):\n", X_T)

# c. Calculate matrix product X.T @ X
XTX = X_T @ X
print("\nc) Matrix Product (X.T @ X):\n", XTX)

# d. Calculate matrix inverse of (X.T @ X)
XTX_inv = np.linalg.inv(XTX)
print("\nd) Inverse of (X.T @ X):\n", XTX_inv)

# e. Implement Ordinary Least Squares (OLS): beta = (X.T @ X)^(-1) @ X.T @ y
beta = XTX_inv @ X_T @ y
print("\ne) OLS Coefficients (beta):\n", beta)

# g. Predict assistance score for new_user = [5, 40, 7]
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(f"\ng) Predicted Assistance Score for new user {new_user}:")
print(f"   Score = {predicted_score:.2f}")
```
