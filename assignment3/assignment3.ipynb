## Q6. Create a sample dataset (employees.csv) and answer the following

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Dana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 55000, 65000, 55000, 60000],
    "Years_of_Experience": [3, 8, 10, 2, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-05-01", "2021-02-10", "2010-11-23"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6500, 4500, 5000],
    "Rating": [4.5, 4.0, 4.8, 4.7, 3.9]
}

emp_df = pd.DataFrame(employee_data)
emp_df.to_csv("employees.csv", index=False)
emp_df
```

**a) Shape (number of rows and columns) of the DataFrame**

```python
emp_df.shape
```

**b) Summary of the DataFrame that includes the data types and non-null counts for each column**

```python
emp_df.info()
```

**c) Descriptive statistics**

```python
emp_df.describe()
```

**d) Display the first 5 and last 3 rows of the dataset**

```python
print("First 5 rows:")
display(emp_df.head())

print("Last 3 rows:")
display(emp_df.tail(3))
```

**e) Calculate the following statistics from the dataset**

```python
avg_salary = emp_df["Salary"].mean()
total_bonus = emp_df["Bonus"].sum()
youngest_employee = emp_df.loc[emp_df["Age"].idxmin(), "Name"]
highest_rated_employee = emp_df.loc[emp_df["Rating"].idxmax(), "Name"]

print("i.   Average salary of employees:", avg_salary)
print("ii.  Total bonus paid to all employees:", total_bonus)
print("iii. Youngest employee:", youngest_employee)
print("iv.  Highest performance rating employee:", highest_rated_employee)
```

**f) Sort the DataFrame by the Salary column in descending order**

```python
emp_df_sorted = emp_df.sort_values(by="Salary", ascending=False)
emp_df_sorted
```

**g) Add a new column that categorizes employees based on their performance rating**

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize_rating)
emp_df
```

**h) Identify missing values in the DataFrame**

```python
emp_df.isnull().sum()
```

**i) Rename the Employee_ID column to ID**

```python
emp_df = emp_df.rename(columns={"Employee_ID": "ID"})
emp_df.columns
```

**j) Find all employees who have more than 5 years of experience AND belong to the IT department**

```python
experienced_it_employees = emp_df[(emp_df["Years_of_Experience"] > 5) & (emp_df["Department"] == "IT")]
experienced_it_employees
```

**k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary**

```python
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df
```

**l) Save the modified DataFrame (with added columns) to a new CSV file**

```python
emp_df.to_csv("employees_modified.csv", index=False)
print("Saved to employees_modified.csv")
```
