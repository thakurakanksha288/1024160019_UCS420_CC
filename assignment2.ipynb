## Q1. Lists

```python
roll_number = "1024160019"

# Extract individual digits and multiply each by 10 to form list L
digits = [int(d) for d in roll_number]
L = [d * 10 for d in digits]

L
```

**i. Print L**

```python
print(L)
```

**ii. Add two numbers — one using append(), one using insert()**

```python
L.append(100)  # append() adds the element 100 to the END of the list
print("After append(100):", L)

L.insert(3, 55)  # insert(3, 55) places 55 at index 3, shifting later elements right by one
print("After insert(3, 55):", L)
```

**iii. Remove two elements — one using remove(), one using pop()**

```python
L.remove(55)  # remove() deletes the FIRST occurrence of the given VALUE (55)
print("After remove(55):", L)

L.pop(0)  # pop(0) deletes the element at the given INDEX (0) and also returns it
print("After pop(0):", L)
```

**iv. Sort ascending, then descending**

```python
L.sort()  # sorts the list in place, ascending order
print("Sorted ascending:", L)

L.sort(reverse=True)  # sorts the list in place, descending order
print("Sorted descending:", L)
```

**v. Slicing — first three and last three elements**

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

**vi. List comprehension — elements greater than the average of L**

```python
avg_L = sum(L) / len(L)
above_avg = [x for x in L if x > avg_L]
print("Average of L:", avg_L)
print("Elements greater than average:", above_avg)
```

## Q2. Tuples

```python
# Using the ORIGINAL L from Q1 (before append/insert/remove/pop/sort)
# to keep this section independent and reproducible.
original_L = [d * 10 for d in digits]
scores = tuple(original_L[:8])
scores
```

**i. Highest score & its index, lowest score & its count**

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"Highest score = {highest} at index {highest_index}")
print(f"Lowest score = {lowest}, appears {lowest_count} time(s)")
```

**ii. Reverse the tuple, returned as a list**

> Tuples are immutable, so there is no in-place `reverse()` method for them; `reversed()` must be used and the result converted to a list (or another tuple).

```python
reversed_scores = list(reversed(scores))
print("Reversed scores (as list):", reversed_scores)
```

**iii. Ask user to input a score, find its first occurrence index**

```python
while True:
    user_input = input("Enter a score to search for: ").strip()
    if not user_input:
        print("Please enter a score.")
        continue
    try:
        user_score = int(user_input)
        break
    except ValueError:
        print("Invalid input. Please enter an integer score.")

if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in scores")
if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in scores")
```

**iv. Attempt to modify a tuple element directly**

> Tuples are immutable, so item assignment raises a `TypeError`; a list would allow this same operation without any error.

```python
print("Attempting scores[0] = 100 ...")
try:
    scores[0] = 100
except TypeError as e:
    print("Error raised:", e)
```

**v. Unpack into first, second, and remaining scores**

```python
first_score, second_score, *remaining_scores = scores
print("first_score =", first_score)
print("second_score =", second_score)
print("remaining_scores =", remaining_scores)
```

## Q3. Random Numbers

```python
random.seed(1024160019)  # seeded with roll number for reproducible-but-unique output
```

**i. Generate 100 random numbers between 100 and 900 (inclusive)**

```python
random_list = [random.randint(100, 900) for _ in range(100)]
print(random_list)
```

**ii. Count odd numbers**

```python
odd_numbers = [n for n in random_list if n % 2 != 0]
print("Count of odd numbers:", len(odd_numbers))
```

**iii. Count even numbers**

```python
even_numbers = [n for n in random_list if n % 2 == 0]
print("Count of even numbers:", len(even_numbers))
```

**iv. Count and list prime numbers**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [n for n in random_list if is_prime(n)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

**v. Most frequently occurring number**

```python
most_common = max(set(random_list), key=random_list.count)
print("Most frequent number:", most_common, "| Occurs:", random_list.count(most_common), "times")
```

## Q4. Sets

```python
# Using the first 8 digits of the roll number (before multiplying by 10)
first_8_digits = digits[:8]

A = {d * 7 for d in first_8_digits}
B = {d * 9 for d in first_8_digits}
print("Set A (digit x 7):", A)
print("Set B (digit x 9):", B)
```

**vi. Union**

```python
print("Union (A | B):", A.union(B))
```

**vii. Intersection**

```python
print("Intersection (A & B):", A.intersection(B))
```

**viii. A − B and B − A**

> `difference()` gives elements in one set only (one direction at a time), while `symmetric_difference()` gives all elements that are in exactly one of the two sets combined (i.e. union minus intersection).

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

**ix. Symmetric difference**

```python
print("Symmetric difference (A ^ B):", A.symmetric_difference(B))
```

**x. Subset / superset checks**

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi. Discard a user-input value from A**

> `discard()` removes X if present but does NOT raise an error if X is missing, unlike `remove()`, which raises a `KeyError` — safer when you're not certain the value exists in the set.

```python
x_value = int(input("Enter a value X to discard from set A: "))
A.discard(x_value)
print("Set A after discard:", A)
```

## Q5. Dictionaries

```python
my_dict = {
    "name": "Akanksha",
    "roll_no": "1024160019",
    "branch": "CSE",
    "age": 20,
    "city": "Jalandhar",
}
my_dict
```

**i. Rename \"city\" to \"location\" (generic approach — works for any dict)**

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

**ii. Add a new key \"cgpa\"**

```python
my_dict["cgpa"] = 8.4
print(my_dict)
```

**iii. Update \"age\" by increasing it by 1**

```python
my_dict["age"] += 1
print(my_dict)
```

**iv. Delete \"branch\" using pop() in one copy, and del in another copy**

> `pop()` removes the key AND returns its value, so it can be captured in a variable, while `del` simply deletes the key-value pair and returns nothing.

```python
dict_pop_version = my_dict.copy()
popped_value = dict_pop_version.pop("branch")
print("pop() returned:", popped_value)
print("Dict after pop():", dict_pop_version)

dict_del_version = my_dict.copy()
del dict_del_version["branch"]
print("Dict after del:", dict_del_version)
```

**v. Iterate using .items()**

```python
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

**vi. Safely check for \"email\" key before accessing it**

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Key 'email' not found - no email on record.")
```

**vii. Merge with a friend's dictionary using `{**dict1, **dict2}`**

> When both dictionaries share a key, the value from the SECOND dictionary (`friend_dict` here, since it comes last in the unpacking order) wins and overwrites the first.

```python
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024160099",
    "branch": "Computer Science",
    "age": 21,
    "city": "Delhi",
}
merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
```

**viii. Dictionary comprehension — keep only string-valued pairs**

```python
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_only_dict)
```
