```python
import numpy as np
#question-1
# Create a 1D NumPy array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

# a) Addition of 2 to all elements
add_result = arr + 2
print("a) Addition of 2:", add_result)

# b) Multiply 3 with all elements
mult_result = arr * 3
print("b) Multiplication by 3:", mult_result)

# c) Divide every element by 2
div_result = arr / 2
print("c) Division by 2:", div_result)
```

```python
#question-2
# --- a) Reverse the NumPy array ---
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]

print("a) Reversed Array:", reversed_arr)

# --- b) Find most frequent value and indices ---

# i. Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[counts_x == counts_x.max()]
indices_x = np.where(x == most_freq_x[0])[0]

print(f"b.i) Array x - Most frequent value: {most_freq_x[0]}")
print(f"     Indices: {indices_x.tolist()}")

# ii. Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
most_freq_y = vals_y[counts_y == counts_y.max()]

print("b.ii) Array y - Most frequent value(s):")
for val in most_freq_y:
    indices_y = np.where(y == val)[0]
    print(f"      Value {val} -> Indices: {indices_y.tolist()}")
```

```python
#question3
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

# a) Access 1st row, 2nd column (Row index 0, Column index 1)
elem_a = arr[0, 1]

# b) Access 3rd row, 1st column (Row index 2, Column index 0)
elem_b = arr[2, 0]

print("a) 1st row, 2nd column:", elem_a)
print("b) 3rd row, 1st column:", elem_b)
```

```python
#question-4
# Create 1-D array with 25 evenly spaced numbers from 10 to 100
alex = np.linspace(10, 100, 25)

# Array attributes
print("Array:", alex)
print("Dimensions (ndim):", alex.ndim)
print("Shape:", alex.shape)
print("Total elements (size):", alex.size)
print("Data type (dtype):", alex.dtype)
print("Total bytes consumed (nbytes):", alex.nbytes)

# Transpose using reshape() (converting 1D row vector to 2D column vector)
alex_transposed = alex.reshape(25, 1)
print("\nTransposed shape using reshape(25, 1):\n", alex_transposed.shape)

# Checking if .T attribute works the same on a 1D array
print("Shape using .T attribute:", alex.T.shape)
```

```python
#question-5
# 1. Create a 3x4 2-D Array
values = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]
ucs420_alex = np.array(values).reshape(3, 4)

print("Original Array (ucs420_alex):\n", ucs420_alex)

# 2. Compute statistical operations
print("\n--- Statistics ---")
print("Mean:", np.mean(ucs420_alex))
print("Median:", np.median(ucs420_alex))
print("Max:", np.max(ucs420_alex))
print("Min:", np.min(ucs420_alex))
print("Unique Elements:", np.unique(ucs420_alex))

# 3. Reshape array to 4 rows and 3 columns
reshaped_ucs420_alex = ucs420_alex.reshape(4, 3)
print("\nReshaped Array (4x3):\n", reshaped_ucs420_alex)

# 4. Resize array to 2 rows and 3 columns
resized_ucs420_alex = np.resize(ucs420_alex, (2, 3))
print("\nResized Array (2x3):\n", resized_ucs420_alex)
```
